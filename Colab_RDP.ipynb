# 🖥️ Google Colab Desktop GUI qua Chrome Remote Desktop

**Cấu hình:** 12.7 GB RAM, CPU Intel Xeon, Mạng Google 1 Gbps siêu mượt 60 FPS.

### 📌 Hướng dẫn 3 bước:
1. Vào link: **https://remotedesktop.google.com/headless**
2. Bấm **Bắt đầu (Begin)** ➔ **Tiếp theo (Next)** ➔ **Ủy quyền (Authorize)**.
3. Copy dòng lệnh ở mục **Debian Linux** dán vào ô `CRP` bên dưới, chọn mã PIN rồi bấm nút **Run (Play)** ▶️!

```python
#@title 🚀 Bấm nút Play ▶️ ở đây để Khởi Động Máy Ảo
#@markdown Dán mã lệnh Debian Linux lấy từ remotedesktop.google.com/headless vào đây:
CRP = "" #@param {type:"string"}
#@markdown Nhập mã PIN gồm 6 số (dùng để mở khóa màn hình trên điện thoại):
Pin = 123456 #@param {type:"integer"}

import os, subprocess, time

if not CRP:
    print("❌ LỖI: Vui lòng dán mã CRP lấy từ https://remotedesktop.google.com/headless vào ô trên!")
elif len(str(Pin)) < 6:
    print("❌ LỖI: Mã Pin phải có ít nhất 6 chữ số!")
else:
    print("⏳ [1/4] Đang tạo tài khoản người dùng...")
    username = "khang"
    os.system(f"useradd -m {username}")
    os.system(f"adduser {username} sudo")
    os.system(f"echo '{username}:KhangSMP@2026' | sudo chpasswd")
    os.system("sed -i 's/\/bin\/sh/\/bin\/bash/g' /etc/passwd")

    print("⏳ [2/4] Đang cài đặt Chrome Remote Desktop...")
    os.system("apt update -qq")
    subprocess.run(['wget', '-q', 'https://dl.google.com/linux/direct/chrome-remote-desktop_current_amd64.deb'], stdout=subprocess.PIPE)
    subprocess.run(['dpkg', '--install', 'chrome-remote-desktop_current_amd64.deb'], stdout=subprocess.PIPE)
    subprocess.run(['apt', 'install', '-qq', '--assume-yes', '--fix-broken'], stdout=subprocess.PIPE)

    print("⏳ [3/4] Đang cài đặt giao diện Desktop (XFCE4) & Trình duyệt Chrome...")
    os.environ['DEBIAN_FRONTEND'] = 'noninteractive'
    os.system("apt install -qq --assume-yes xfce4 desktop-base xfce4-terminal xscreensaver > /dev/null 2>&1")
    os.system("bash -c 'echo \"exec /etc/X11/Xsession /usr/bin/xfce4-session\" > /etc/chrome-remote-desktop-session'")
    os.system("systemctl disable lightdm.service > /dev/null 2>&1")

    subprocess.run(['wget', '-q', 'https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb'], stdout=subprocess.PIPE)
    subprocess.run(['dpkg', '--install', 'google-chrome-stable_current_amd64.deb'], stdout=subprocess.PIPE)
    subprocess.run(['apt', 'install', '-qq', '--assume-yes', '--fix-broken'], stdout=subprocess.PIPE)

    print("⏳ [4/4] Đang kết nối vào tài khoản Google...")
    os.system(f"adduser {username} chrome-remote-desktop > /dev/null 2>&1")
    command = f"{CRP} --pin={Pin}"
    os.system(f"su - {username} -c '{command}'")
    os.system("service chrome-remote-desktop start")

    print("\n===========================================================")
    print("🎉 MÁY ẢO GOOGLE COLAB ĐÃ SẴN SÀNG 100%!")
    print("📱 Mở app Chrome Remote Desktop trên điện thoại hoặc vào web:")
    print("👉 https://remotedesktop.google.com/access")
    print(f"🔑 Nhập mã PIN: {Pin}")
    print("===========================================================\n")

    # Giữ máy ảo không bị tắt
    while True:
        time.sleep(300)
```
